###Laboratorio 4: Porposiciones

###**Objetivos:** Un sistema inteligente supervisa una red empresarial y debe determinar si existe una condición crítica de servicio
**Fecha:** 7 de septiembre

## Preposiciones 

P: El servidor no esta disponible
Q: Existe latencia
R: Existe perdidda de paquetes alta
S: El tráfico es anormalmente alto
T: El servidor de respaldo no esta disponible
U: El sistema de monitoreo esta activado

```python
## Decalración de Variables
servidor = input("¿El servidor no esta disponible? (s/n): ").lower() =="n"
latencia = input("¿La latencia supera el limite establecido? (s/n): ").lower() =="s"
paquetes = input("¿La perdida de paquetes supera el limite? (s/n): ").lower() =="s"
trafico = input("¿El trafico de red es mayor de lo habitual? (s/n): ").lower() =="s"
servidorRespaldo= input ("¿El servidor de respaldo no esta disponible? (s/n): ").lower() =="n"
monitoreo = input ("¿El sistema de monitoreo esta activo? (s/n): ").lower() =="s"

critico=(servidor and latencia and paquetes and trafico and servidorRespaldo and monitoreo)
print ("Revisión.....")
if critico == True:
    print("Alerta el nivel de condición es critico")
else:
    print("El nivel de condición aun es optimo para trabajar")
```
