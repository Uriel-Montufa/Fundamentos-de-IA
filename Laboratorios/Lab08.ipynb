```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("Servidor_1","RouterA"),
        ("RouterA","SwitchA"),
        ("RouterA","SwitchB"),
        ("SwitchB","PC2"),
        ("SwitchB","ImpresoraA"),
        ("SwitchA","ImpresoraA"),
        ("SwitchA","PC1"),
    ])
```

```python
nx.draw(
    grafo,
    with_labels=True,
    node_size=1000,
    node_color="#AAA",
    node_shape="o"
)
plt.show()
```
