```python
#Limpieza de los datos
#Dato nulo es aquel que no existe o no se conoce
titanic.isnull().sum()
```

```python
#LLenamos los valores nulos con la mediana de la edad
titanic["Age"] = titanic["Age"].fillna(
    titanic["Age"].median()
)
```

```python
titanic.isnull().sum()
```

```python
#¿Cuantos murieron?
titanic["Survived"].value_counts()
```

```python
#¿Que voy a predecir?
#Si alguien puede sobrevivir o no
columnas =[
    "Pclass",
    "Sex",
    "Age",
    "Fare"
]
```

```python
#Variables predictoras
X= titanic[columnas]
#Variables a predecir / objetivo
y= titanic["Survived"]
```

```python
X.columns
```

```python
y.name
```

```python
#Validamos la existencia de valores nulos
X.isnull().sum()
```

```python
#Lo llenamos con el valor más repetido
X["Fare"] = X["Fare"].fillna(
    X["Fare"].mode()[0]
)
```

```python
X.isnull().sum()
```

## Normalización de los datos.

```python
#Los algoritmos de Machine Learning no entienden texto.
```

```python
X.head(2)
```

```python
# One Hot Encoding: Permite convertir variables de clase en numeros
X= pd.get_dummies(
    X,
    columns=["Sex"],
    drop_first=True
)
```

```python
X.head(3)
```

## División de los Datos

Datos para entrenar (train) - 70%

Datos para prueba (test) - 30%

```python
#sklearn: Libreria de machine learning para Phyton
# Random Forest
# Arbol de decision
# KNN
# Baige Naives
# Regresion
```

```python
#Metodo para dividir los datos
from sklearn.model_selection import train_test_split
```

```python
#Metodo para escalar los datos
from sklearn.preprocessing import StandardScaler
```

```python
#Algoritmo Arbol de desición
from sklearn.tree import DecisionTreeClassifier
```

```python
#Dividamos los datos
X_train, X_test, y_train, y_test = train_test_split(
    X, #Variables predictoras
    y, #Variables objetivo
    test_size = 0.20, #Porcentaje para los datos de prueba
    random_state=42 #Semilla, para garantizar la reproducibilidad
)
```

```python
print("Entrenamiento datos predictorias (X): ", X_train.shape)
```

```python
print("Entrenamiento datos predictorias (y): ", y_train.shape)
```

```python
print("Prueba datos predictorias (X): ", X_test.shape)
```

```python
print("Prueba datos predictorias (y): ", y_test.shape)
```

## Escalado de los datos

```python
columnas_numericas =["Age","Fare"]
```

```python
X_train[columnas_numericas].describe()
```

```python
# Escalamos ls datos usando StandarScaler
scaler = StandardScaler()
```

```python
# Escalamos las variables numericas
#Dataset de entrenamiento
X_train[columnas_numericas] = scaler.fit_transform(X_train[columnas_numericas])
#Dataset de prueba
X_test[columnas_numericas] = scaler.transform(X_test[columnas_numericas])
```

```python
# Mostrar datos escalados
X_train[columnas_numericas].describe()
```

# Entrenamiento del Algoritmo
## Arbol de decisión: Permite clasificar los registros con respecto a las variables

```python
#Declaramos el modelo y su configuración
modelo = DecisionTreeClassifier(
    max_depth=4, #La profundidad del arbol
    random_state=42 #La semilla paar garantiazr la reproducibilidad
)
```

```python
#Entrenamos el modelo
modelo.fit(X_train, y_train)
```

## Evaluar el Modelo
## Vamos a evaluar la eficiencia del modelo, usando el conjunto de pruebas

```python
# Importamos librerias
from sklearn.metrics import accuracy_score
```

```python
predicccion = modelo.predict(
    X_test
)
```

```python
pd.DataFrame(
    {
        "Survived [Valor Real]": y_test,
        "Survived [Prediccion]": predicccion
    }
).head(20)
```

```python
exactitud = accuracy_score(y_test, predicccion)
```

```python
print(f"La exactitud del modelo es: {exactitud: .4f}")
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, predicccion))
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test, predicccion)
print("Matriz de confusion")
print(cm)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
matriz = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[0,1]
)
matriz.plot(
    cmap = plt.cm.Blues
)
plt.show
```
