```python
#Importar las librerias
import matplotlib.pyplot as plt #Librerias para graficar
import networkx as nx #Libreria que ne permite trabajar con estructuras complejas, como Grafos
```

```python
#Crear una instancai de la clase Graph
#Graph es una colección de nodos 
grafo = nx.Graph()
```

```python
#Creamos el grafo utilizandoel método add_edges
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("A","E"),
        ("B","D"),
        ("D","F"),
        ("E","G"),
        ("E","F")
    ]
)
```

```python
#Mostrar el grafo
nx.draw(
    grafo,
    with_labels=True,
    node_size=1000,
    node_color="#AAA",
    node_shape="o"
)
plt.show()
```

```python
#Mostrar las rutas posibles del grafo
#Source = De donde inicia
#Target = El objetivo
rutas = list (
    nx.all_simple_paths(grafo,source="A", target="G")
)
```

```python
#Imprimir las rutas
print ("Rutas encontradas: ")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Mostrar el total de pasos de cada ruta
pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1: ",pasos1)
pasos2 = len(rutas[1])-1
print("pasos de la ruta 2: ",pasos2)
```

```python
#Aplicamos la lógica proposicional
P = pasos1 < pasos2 # P: La mejor ruta es la 1
Q = pasos1 > pasos2 # Q: La mejor ruta es la 2
if P:
    print("P = True")
    print("La ruta 1 es la mejor opción.")
elif Q: 
    print("Q = True")
    print("La ruta 2 es la mejor opción.")
else:
    print("P = False")
    print("Q = False")
    print("Las dos rutas son igual de buenas, haz lo que quieras")
```
