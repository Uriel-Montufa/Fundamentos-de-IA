```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
#Creamos el grafo
grafo = nx.DiGraph()
```

```python
grafo.add_weighted_edges_from(
    [
        ("A","B",5),
        ("A","C",2),
        ("B","D",4),
        ("C","E",5),
        ("C","F",6),
    ])
```

```python
#Imprimimos el grafo
pos = nx.spring_layout(grafo, seed=10)
nx.draw(
    grafo,
    pos,
    with_labels= True,
    node_size=1500,
    arrows=True
)
labels= nx.get_edge_attributes(grafo, "weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#¿Puedo moverme de A a B?
print (grafo.has_edge ("A","B"))
```

```python
#¿Puedo moverme de B a A?
print (grafo.has_edge ("B","A"))
```

```python
#¿Puedo moverme de A a F?
print (grafo.has_edge ("A","F"))
```

```python
#Mostramos las rutas
rutas = list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="F"
    )
)
```

```python
#Imprimiimos las rutas
for i, ruta in enumerate (rutas, start=1):
    print (f"Ruta {i}:{ruta} ")
```

```python
grafo2 =nx.DiGraph()
```

```python
grafo2.add_weighted_edges_from(
    [
        ("A","B",4),
        ("B","D",3),
        ("D","E",2),
        ("B","E",6),
        ("C","D",2),
        ("A","C",7),
        ("C","F",5),
        ("F","G",4),
        ("D","G",8),
        ("E","G",3),
    ])
```

```python
#Imprimimos el grafo
pos = nx.spring_layout(grafo2, seed =12)
nx.draw(
    grafo2,
    pos,
    with_labels= True,
    node_size=1500,
    arrows=True
)
labels= nx.get_edge_attributes(grafo2, "weight")
nx.draw_networkx_edge_labels(
    grafo2,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#Mostramos las rutas
rutas2 = list(
    nx.all_simple_paths(
        grafo2,
        source="A",
        target="G"
    )
)
```

```python
#Imprimiimos las rutas
for i, ruta2 in enumerate (rutas2, start=1):
    print (f"Ruta {i}:{ruta2} ")
```

```python
#Calculamos el costo
def costo_ruta(grafo2, ruta2):
    costo=0
    for i in range(len(ruta2)-1):
        costo += grafo2 [ruta2[i]][ruta2[i+1]]["weight"]
    return costo
```

```python
#Imprimimos el costo
for i, ruta2 in enumerate (rutas2, start=1):
    costo = costo_ruta (grafo2,ruta2)
    print (f"Ruta {i}:{ruta2}->{costo} unidades")
```

```python
#Elegir la ruta mas corta
costos=[ costo_ruta(grafo2,ruta2)
         for ruta2 in rutas2
       ]
mejor_ruta= costos.index(min(costos))
```

```python
#Imprimir la mejor ruta
print(f"La mejor ruta es:{mejor_ruta+1}")
print(f"El mejor camino corto es:{costos[mejor_ruta]}")
```
