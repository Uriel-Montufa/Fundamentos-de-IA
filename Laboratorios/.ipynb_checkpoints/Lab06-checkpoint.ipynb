### Laboratorio 6: Grafos
# Descripción: Encuentra la ruta
# Situación: Un agente inteligente se encuentra en A y necesita llegar a J.

```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("B","D"),
        ("C","G"),
        ("D","F"),
        ("F","H"),
        ("H","I"),
        ("I","J"),
        ("F","E"),
        ("G","E"),
        ("G","J"),
    ])
```

```python
nx.draw(
    grafo,
    with_labels=True,
    node_size=1000,
    node_color="#AAA",
    node_shape="o"
)
plt.show()
```

```python
rutas = list (
    nx.all_simple_paths(grafo,source="A", target="J")
)
```

```python
print ("Rutas encontradas: ")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
pasos1 = len(rutas [0])-1
print("Pasos en la ruta 1: ",pasos1)
pasos2 = len(rutas [1])-1
print("Pasos en la ruta 2: ",pasos2)
#pasos3 = len(rutas [2])-1
#print("Pasos en la ruta 3: ",pasos3)
#pasos4 = len(rutas [3])-1
#print("Pasos en la ruta 4: ",pasos4)
```

```python
P = pasos1 < pasos2 #and pasos1 < pasos3 and pasos1 < pasos4 # P: La mejor ruta es la 1
Q = pasos1 > pasos2 #and pasos2 < pasos3 and pasos2 < pasos4 # Q: La mejor ruta es la 2
#R = pasos1 > pasos3 and pasos3 < pasos2 and pasos3 < pasos4 # R: La mejor ruta es la 3
#S = pasos1 > pasos4 and pasos4 < pasos3 and pasos4 < pasos2 # S: La mejor ruta es la 4
if P:
    print("P = True")
    print("La ruta 1 es la mejor opción.")
elif Q: 
    print("Q = True")
    print("La ruta 2 es la mejor opción.")
#elif R: 
    #print("R = True")
    #print("La ruta 3 es la mejor opción.")
#elif S: 
   # print("S = True")
    #print("La ruta 4 es la mejor opción.")
else:
    print("P = False")
    print("Q = False")
    #print("R = False")
    #print("S = False")
    print("Las cuatro rutas son igual de buenas, haz lo que quieras")
```
