```python
### Laboratorio 11: Grafos Dirigidos y Ponderados
# Algoritmo dijkstra
```

```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
grafo = nx.DiGraph()
```

```python
grafo.add_weighted_edges_from(
    [
        ("A","B",8),
        ("A","D",5),
        ("A","E",4),
        ("B","C",3),
        ("B","F",4),
        ("C","F",9),
        ("C","G",2),
        ("F","K",8),
        ("K","J",6),
        ("J","O",9),
        ("O","P",12),
        ("E","I",5),
        ("D","E",3),
        ("D","H",6),
        ("H","M",2),
        ("H","I",4),
        ("M","N",1),
        ("N","O",2),
        ("F","E",3),
        ("I","J",4),
        ("I","N",3),
        ("I","M",1),
        ("P","K",1),
        ("G","L",2),
        ("G","K",3),
        ("L","K",5),
        ("L","P",3),
    ])
```

```python
pos = nx.spring_layout(grafo, seed=12)
nx.draw(
    grafo,
    pos,
    with_labels= True,
    node_size=1500,
    arrows=True
)
labels= nx.get_edge_attributes(grafo, "weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#Mostramos las rutas
rutas = list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="P"
    )
)
```

```python
#Imprimiimos las rutas
for i, ruta in enumerate (rutas, start=1):
    print (f"Ruta {i}:{ruta} ")
```

```python
#Calculamos el costo
def costo_ruta(grafo, ruta):
    costo=0
    for i in range(len(ruta)-1):
        costo += grafo [ruta[i]][ruta[i+1]]["weight"]
    return costo
```

```python
#Imprimimos el costo
for i, ruta in enumerate (rutas, start=1):
    costo = costo_ruta (grafo,ruta)
    print (f"Ruta {i}: {ruta}-> {costo}")
```

```python
#Elegir la ruta mas corta
costos=[ costo_ruta(grafo,ruta)
         for ruta in rutas
       ]
mejor_ruta= costos.index(min(costos))
```

```python
#Imprimir la mejor ruta
print(f"La mejor ruta es:{mejor_ruta+1}")
print(f"El mejor camino corto es:{costos[mejor_ruta]}")
```

```python
#Elegir la ruta mas larga
costos=[ costo_ruta(grafo,ruta)
         for ruta in rutas
       ]
peor_ruta= costos.index(max(costos))
```

```python
#Imprimir la mejor ruta
print(f"La peor ruta es:{peor_ruta+1}")
print(f"El peor camino corto es:{costos[peor_ruta]}")
```
