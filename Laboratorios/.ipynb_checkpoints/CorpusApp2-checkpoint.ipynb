```python
# 1. IMPORTAR LIBRERÍAS Y CARGAR DATOS
import pandas as pd

# Cargar el archivo original
dfRehab = pd.read_csv('rehabilitation_movement_dataset.csv')

# 2. DEFINIR LAS VARIABLES A DESCARTAR
# Escribe aquí exactamente los nombres de las columnas que mostraron 
# baja importancia o alta redundancia en tus gráficos anteriores.
variables_a_descartar = [
    'Quaternion_z', 
    'Quaternion_w', 
    'Age (yrs)',
    'Quaternion_x',
    'Accel_Y (m/s²)'
]

# 3. ELIMINAR LAS VARIABLES DEL DATAFRAME
df_limpio = dfRehab.drop(columns=variables_a_descartar)

# Verificar que se hayan eliminado correctamente
print(f"Columnas originales: {dfRehab.shape[1]}")
print(f"Columnas tras el descarte: {df_limpio.shape[1]}")

# 4. GUARDAR EL NUEVO DATASET (Recomendado)
# Exportar este nuevo conjunto a un CSV para que tus futuros modelos 
# se entrenen directamente con este archivo limpio.
df_limpio.to_csv('rehabilitation_dataset_optimizado.csv', index=False)
print("\n¡Dataset optimizado guardado con éxito!")
```
