```python
###Laboratorio 3: Porposiciones

###**Objetivos:** Un videojuego utliliza un agente inteligente enemigo que debe decidir si puede iniciar un ataque contra el jugador
### Iniciar un ataque contra el jugador
**Fecha:** 7 de septiembre
```

#Proposiciones
P: Se encuentra dentro del rango
Q: El nivel de vida es suficiente
R: Existe municiones disponibles
S: Esta visible
T: No existe amenaza cercana
U: No esta aturdido

```python
### Definición de variables

rango = True # Si esta dentro del rango
nivelVida = True # Si tiene vida suficiente 
municiones = True # Si ceunta con municiones disponibles
visible = True # Si esta visible el jugador
amenaza = False # No existe amenaza cercana
aturdido= False # No esta aturdido
```

```python
puedeAtacar = (rango==True and visible==True and nivelVida==True and municiones==True 
               and amenaza==False and aturdido==False)
```

```python
print ("¿El agente puede atacar?", puedeAtacar)
```

```python
print (" ==Agente Enemigo Inteligente==")
print ("Development by Uriel")
rango = input ("El jugador se encuentra dentro del rango (s/n): ").lower()=="s"
visible = input("El jugador esta visible (s/n): ").lower()=="s"
nivelVida = input ("Tienes suficiente vida (s/n): ").lower()=="s"
municiones = input ("El enemigo tiene munición (s/n): ").lower()=="s"
amenaza = input ("Existe una amenaza cercana (s/n): ").lower()=="n"
aturdido = input ("El enemigo no esta aturdido (s/n): ").lower()=="n"

puedeAtacar = (rango and visible and nivelVida and municiones and amenaza and aturdido)
print ("Desición del agente")
if puedeAtacar == True :
    print("El agente puede atacar: en marcha ")
else:
    print("El agente no puede atacar: a cubierto")
```
