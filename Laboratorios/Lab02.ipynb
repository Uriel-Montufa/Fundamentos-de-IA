# Problema: Realizar un sistema que identifique si una planta necesita riego para activar el sensor.

```python
print("Activación de riego")
humedad = float(input("Nivel de Humedad 0%-100%: "))
```

```python
print("Valor almacenado")
print ("Humedad: ",humedad, "%")
```

```python
if humedad>=40 and humedad<=100:
    print("La planta esta llena de vida no necesita agua")
elif humedad<40 and humedad>=0:
    print("La planta necesita atención, riegala")
else:
    print("Valor invalido")
```
